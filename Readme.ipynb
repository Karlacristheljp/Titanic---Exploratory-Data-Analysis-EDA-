Titanic Exploratory Data Analysis (EDA)

This repository contains a comprehensive, academic-grade Exploratory Data Analysis (EDA) on the Titanic Dataset (train.csv). The project investigates the socio-demographic, economic, and age-related drivers of passenger survival across 891 observations and 12 features.

**Analytical Workflow & Notebook Structure**

The analysis in Src/eda_titanic_notebook.ipynb is structured into five core diagnostic phases:


**1. Structure & Dimension Audit**

- Loads train.csv relative to the Src/ folder using Python's os.path.join("..", "Data", "train.csv").

- Confirms dataset dimensions of 891 rows × 12 columns (df_titanic.shape).

- Audits data types via df_titanic.info()

- Generates primary summary statistics using df_titanic.describe().T.


**2. Missing Values Analysis**

- Quantifies null counts using df_titanic.isnull().sum().

- Identifies features with missing data: Cabin (687 missing, 77.10%), Age (177 missing, 19.87%), and Embarked (2 missing, 0.22%).



**3. Univariate & Bivariate Analysis: Hypotheses Testing**

- Hypothesis #1: Passenger Class (Pclass) & Survival: Evaluates survival rates across socio-economic tiers. Class 1 leads with 62.96%, Class 2 at 47.28%, and Class 3 drops to 24.23%. When disaggregated by gender, Class 1 females reached a 96.80% survival rate compared to 36.88% for Class 1 males. Saved to ../Results/hypothesis_class_survival_percentage.png and ../Results/hypothesis_class_survival_and sex.png.

- Hypothesis #2: Gender (Sex) & Survival: Demonstrates extreme discrepancy between demographic groups. Females achieved an overall survival rate of 74.20%, against 18.89% for males, confirming gender as the primary predictor. Saved to ../Results/hypothesis_survival_rates_by_gender.png.

- Hypothesis #3: Age (Age) Distribution & Survival: Evaluates non-normality and age cohorts using subplots with Kernel Density Estimation (sns.histplot(kde=True)) and stacked count histograms. Reveals non-linear peaks favoring infants/children (ages 0–5) and young adults (ages 15–35), while fatalities clustered heavily between ages 20 and 40. Saved to ../Results/Age_Distribution_Density.png and ../Results/hypothesis_age_survival.png.



**4. Correlation Matrix & Numerical Drivers**

Filters numerical features (Survived, Pclass, Age, SibSp, Parch, Fare) and calculates the Pearson correlation matrix (.corr()). Top linear associations are rendered in a coolwarm heatmap (../Results/correlation_matrix.png):

- Pclass (r = -0.34): Inverse linear correlation confirming upper-class passengers had a significantly higher probability of survival.
- Fare (r = 0.26): Positive correlation indicating higher ticket fares directly correlate with upper-class placement and survival.
- Pclass vs. Age (r = -0.37): Indicates older passengers predominantly occupied Class 1 accommodations.
- Age (r = -0.08): Demonstrates weak linear correlation, confirming the non-linear relationship analyzed via density distributions.

### How to Reproduce

**1. Prerequisites**

Install Python 3.8+ and the necessary data science libraries:

pip install pandas numpy matplotlib seaborn jupyter

**2. Running the Analysis**

Clone this repository:

git clone https://github.com/Karlacristheljp/Titanic---Exploratory-Data-Analysis-EDA-

Launch JupyterLab or Jupyter Notebook:

jupyter notebook Src/Titanic-EDA_notebook.ipynb


Execute all cells sequentially. The notebook will automatically write all visualization PNGs directly into the Results/ directory.
